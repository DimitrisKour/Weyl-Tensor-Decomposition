## 1. Minkowski (conformally flat) spacetime — brief intro

In Cartesian coordinates $x^\mu=(t,x,y,z)$ the Minkowski metric is
$$
\eta_{\mu\nu}=\mathrm{diag}(-1,\,1,\,1,\,1).
$$

This spacetime is flat: all Christoffel symbols vanish, so the Riemann tensor $R^\rho{}_{\sigma\mu\nu}=0$, the Ricci tensor $R_{\mu\nu}=0$, and the scalar curvature $R=0$.

The Weyl tensor $C_{\alpha\beta\gamma\delta}$, being the trace-free part of the Riemann tensor, also vanishes:
$$
C_{\alpha\beta\gamma\delta}=0.
$$

Because Minkowski space is conformally flat, any metric of the form $g_{\mu\nu}=\Omega^2(x)\,\eta_{\mu\nu}$ is also conformally flat and has $C_{\alpha\beta\gamma\delta}=0$.

Consequences used below: the electric and magnetic parts of the Weyl tensor relative to any timelike observer,
$E_{ab}=C_{acbd}u^c u^d$ and $B_{ab}={}^\star C_{acbd}u^c u^d$, vanish; hence the Bel–Robinson tensor built from $C$ vanishes and the associated scalars/fluxes satisfy $U=P=0$, while curvature invariants based on $C$ give $J_1=J_2=0$.


```python
from utils import *       
from plotting import *    
import sympy as sp
```

```python
# --- 2. define coordinates and metric ---
t, x, y, z = sp.symbols("t x y z", real=True)
coords = (t, x, y, z)

# Minkowski metric (signature -,+,+,+)
eta = sp.diag(-1, 1, 1, 1)

# create EinsteinPy metric
metric = metric_from_components(eta, coords)
metric.tensor()
```

```python
# --- 3. compute Weyl tensor ---
C = weyl_from_metric(metric)

# verify that all components are zero
all_zero = all(val == 0 for val in C.values())
print(f"All Weyl components zero? {all_zero}")
```

```python
# Weyl invariants
W1, W2 = weyl_invariants(C, metric, "raw")
print("W1 =", W1)
print("W2 =", W2)

# Bel–Robinson observer check
u_cov = sp.Matrix([-1, 0, 0, 0])  # static observer
BR = bel_robinson_observer(C, metric, u_cov)
print("U =", BR["U"])
print("P =", BR["P_cov"])
```

### Summary of results for Minkowski / conformally flat

From the computation above we confirmed
$$
C_{\alpha\beta\gamma\delta}=0 \quad\Rightarrow\quad
E_{ab}=0,\; B_{ab}=0,\; J_1=C_{\alpha\beta\gamma\delta}C^{\alpha\beta\gamma\delta}=0,\; J_2=C_{\alpha\beta\gamma\delta}{}^\star C^{\alpha\beta\gamma\delta}=0,
$$
and therefore the Bel–Robinson quantities vanish,
$$
U = T_{\alpha\beta\gamma\delta}u^\alpha u^\beta u^\gamma u^\delta = 0,
\qquad
P_a = -h_a{}^{\mu} T_{\mu\beta\gamma\delta} u^\beta u^\gamma u^\delta = 0.
$$
Since the Weyl tensor captures the conformally invariant, trace-free part of curvature, any metric conformal to Minkowski also has $C_{\alpha\beta\gamma\delta}=0$. 

### Electromagnetic analogue (vacuum)

In flat spacetime, the vacuum EM case $F_{\mu\nu}=0$ implies vanishing field invariants and a zero electromagnetic stress–energy tensor $T^{\mu\nu}_{\rm EM}=0$. More generally, Maxwell’s equations are formulated on Minkowski spacetime in special relativity (no back-reaction in that setting), whereas in general relativity any nonzero electromagnetic field contributes stress–energy and therefore **sources curvature** through the Einstein–Maxwell equations.

**References:**  
Misner, Thorne & Wheeler, *Gravitation*, §5.3–5.6.  
Landau & Lifshitz, *The Classical Theory of Fields*, Ch. 9.  
Jackson, *Classical Electrodynamics* (3rd ed.), §6.7.

---

### Why EM waves on flat space vs. gravitational waves

Electromagnetic waves can propagate on a fixed Minkowski background because Maxwell’s theory is linear and does not require dynamical spacetime geometry. Gravitational waves, by contrast, are perturbations of the spacetime metric itself. In the high-frequency (short-wavelength) limit of GR, gravitational waves carry an **effective stress–energy tensor** (Isaacson), which appears only at **second order** in the perturbation and governs the back-reaction of the waves on the background geometry. This provides the standard gauge-invariant description of gravitational-wave energy transport in vacuum.

**References:**  
Isaacson, R. A., *Phys. Rev.* **166**, 1263–1271 (1968); **167**, 1272–1280 (1968).  
Maggiore, *Gravitational Waves*, Vol. 1 (2008), §2.6.  
Poisson & Will, *Gravity* (2014), §7.2.

### FLRW remark (conformally flat but curved)

FLRW spacetimes are **conformally flat**, meaning the Weyl tensor vanishes:
$$
\Psi_0 = \Psi_1 = \Psi_2 = \Psi_3 = \Psi_4 = 0.
$$
Nevertheless, curvature is present due to the **Ricci tensor**, sourced by matter content and cosmic expansion. Thus, for all spatial curvatures \(k=0, \pm1\), FLRW universes have vanishing Weyl curvature, with dynamics fully encoded in the Ricci part of the curvature decomposition.

**References:**  
Wald, *General Relativity* (1984), §3.2.  
Carroll, *Spacetime and Geometry* (2004), §4.3.  
Hawking & Ellis, *The Large Scale Structure of Space-Time* (1973), §4.3.
