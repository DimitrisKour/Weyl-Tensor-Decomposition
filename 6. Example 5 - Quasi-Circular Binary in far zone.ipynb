### Polarizations (quasi-circular binary, far zone)

For a binary system in a quasi-circular orbit, the leading-order (quadrupole) gravitational-wave polarizations measured by a distant detector are  

$$
\boxed{
\begin{aligned}
h_+(t)
 &= \frac{4}{r}
    \left(\frac{G\,\mathcal M}{c^2}\right)^{5/3}
    \left(\frac{\pi f_{\rm gw}}{c}\right)^{2/3}
    \frac{1+\cos^2\!\mathcal{i}}{2}\,
    \cos\!\big(2\pi f_{\rm gw}t_{\rm ret}+2\phi_0\big),\\[4pt]
h_\times(t)
 &= \frac{4}{r}
    \left(\frac{G\,\mathcal M}{c^2}\right)^{5/3}
    \left(\frac{\pi f_{\rm gw}}{c}\right)^{2/3}
    \cos\!\mathcal{i}\,
    \sin\!\big(2\pi f_{\rm gw}t_{\rm ret}+2\phi_0\big),
\end{aligned}}
$$

where the following parameters characterize the source–detector system:

| Symbol | Description | Type / Meaning |
|:--|:--|:--|
| $r$ | Distance from source to detector (luminosity distance) | Fixed spacetime scale |
| $\mathcal{M}=(\mu^3M^2)^{1/5}$ | Chirp mass of the binary | Source parameter |
| $f_{\rm gw}$ | Gravitational-wave frequency (twice the orbital frequency) | Source parameter |
| $t_{\rm ret}=t-\tfrac{r}{c}$ | Retarded time | Time delay for propagation |
| $\mathcal{i}$ | Inclination angle between orbital angular momentum and line of sight | Orientation parameter |
| $\phi_0$ | Constant orbital phase at reference time | Source phase offset |

Thus, $\mathcal{i}$ and $\phi_0$ are **fixed geometric parameters** of the system, not spacetime coordinates.  
The waveform is circularly polarized for $\mathcal{i}=0$ (face-on) and linearly polarized for $\mathcal{i}=\pi/2$ (edge-on).

---

### Polarization tensors on the celestial sphere

On the unit two-sphere with metric  
$\gamma_{AB}=\mathrm{diag}(1,\sin^2\theta)$  ($A,B\in\{\theta,\varphi\}$),  
the transverse-traceless polarization tensors are

$$
e^+_{AB}=
\begin{pmatrix}
1 & 0\\[2pt]
0 & -\sin^2\theta
\end{pmatrix},
\qquad
e^\times_{AB}=
\begin{pmatrix}
0 & \sin\theta\\[2pt]
\sin\theta & 0
\end{pmatrix},
$$

satisfying $\gamma^{AB}e^{\pm}_{AB}=0$ and $\nabla^A e^{\pm}_{AB}=0$.  
Here $\theta,\varphi$ are the usual **spacetime angular coordinates** describing directions on the sky and are unrelated to the inclination angle $\mathcal{i}$.

---

### Far-zone TT metric in $(t,r,\theta,\varphi)$

Keeping only terms linear in the gravitational-wave amplitudes $h_+$ and $h_\times$,  
the spacetime metric in the radiation zone ($r\gg \lambda_{\rm gw}$) is

$$
ds^2
 = -c^2dt^2 + dr^2
 + r^2\!\left[\gamma_{AB}
   + h_+(t)\,e^+_{AB}
   + h_\times(t)\,e^\times_{AB}\right]
   d\vartheta^A d\vartheta^B,
$$

i.e. in component form,

$$
\boxed{
\begin{aligned}
g_{tt} &= -1,\qquad g_{rr}=+1,\qquad g_{tA}=g_{rA}=0,\\[3pt]
g_{\theta\theta} &= r^2\,[1+h_+(t)],\\[3pt]
g_{\theta\varphi} &= r^2\,h_\times(t)\,\sin\theta,\\[3pt]
g_{\varphi\varphi} &= r^2\,[1-h_+(t)]\,\sin^2\theta,
\end{aligned}}
\qquad \text{(all other first-order components vanish).}
$$

The corresponding metric tensor in matrix form, written in the coordinate basis
$(t,\,r,\,\theta,\,\varphi)$, is therefore

$$
g_{\mu\nu} =
\begin{pmatrix}
-1 & 0 & 0 & 0\\[4pt]
0 & 1 & 0 & 0\\[4pt]
0 & 0 & r^2\!\big(1+h_+(t)\big) & r^2 h_\times(t)\sin\theta\\[4pt]
0 & 0 & r^2 h_\times(t)\sin\theta & r^2\!\big(1-h_+(t)\big)\sin^2\theta
\end{pmatrix}.
$$

At linear order, $h_+(t)$ and $h_\times(t)$ represent the two independent polarization
modes of the outgoing gravitational wave.  
In the limit $h_\pm \to 0$, the metric smoothly reduces to the flat-space spherical form.

### Orbital Angular Velocity

At leading (Newtonian) order, the orbital motion of the binary is governed by Kepler’s third law.  
The orbital angular velocity $\Omega$ is therefore taken to be **Keplerian**, satisfying
$$
\Omega^2 = \frac{G M}{r_0^3},
$$
where
$$
M = m_1 + m_2
$$
is the total mass of the system and $r_0$ is the orbital separation.

This approximation assumes that the gravitational field is weak and the orbital motion slow ($v \ll c$), so that post-Newtonian corrections can be neglected.  
In this regime, $\Omega$ is considered constant in time and directly determines the emitted gravitational-wave frequency
$$
f_{\text{GW}} = \frac{\Omega}{\pi}.
$$
This **Keplerian relation** will be adopted throughout the following calculations.


### Bondi Null Tetrad

The **Bondi null tetrad** in spherical coordinates $(t,r,\theta,\phi)$ is defined as
$$
\ell^a = (0,\,1,\,0,\,0), \qquad
n^a = (1,\,-\tfrac{1}{2},\,0,\,0), \qquad
m^a = \frac{1}{\sqrt{2}r}(0,\,0,\,1,\,\tfrac{i}{\sin\theta}),
$$
and
$$
\bar{m}^a = \frac{1}{\sqrt{2}r}(0,\,0,\,1,\,-\tfrac{i}{\sin\theta}).
$$

They satisfy
$$
\ell\!\cdot\!n = -1, \qquad
m\!\cdot\!\bar{m} = 1, \qquad
\text{all other products vanish.}
$$

In this tetrad, the only nonvanishing Weyl scalar is
$$
\psi_4 = -\tfrac{1}{2}\big(\ddot{h}_+ - i\,\ddot{h}_\times\big),
$$
which encodes the outgoing gravitational radiation.
