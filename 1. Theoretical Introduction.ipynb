# An Introduction to Gravitational Radiation

## Part I: Formalism

## 1. Electrodynamics

## 1.1. Classical Field Description
## 1.1.1 Maxwell's Equations

In their standard vectorial form (in SI units), Maxwell's equations are written as

$$
\begin{aligned}
\nabla \cdot \mathbf{E} &= \frac{\rho}{\varepsilon_0}, \\
\nabla \cdot \mathbf{B} &= 0, \\
\nabla \times \mathbf{E} &= -\frac{\partial \mathbf{B}}{\partial t}, \\
\nabla \times \mathbf{B} &= \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}.
\end{aligned}
$$

Here, 
$\mathbf{E}$ and $\mathbf{B}$ are the electric and magnetic fields, 
$\rho$ is the charge density, and $\mathbf{J}$ is the current density. 
The constants $\varepsilon_0$ and $\mu_0$ denote the vacuum permittivity and permeability respectively.

From these equations, one can derive the **continuity equation**

$$
\frac{\partial \rho}{\partial t} + \nabla \cdot \mathbf{J} = 0,
$$

which expresses the local conservation of electric charge.

## 1.1.2 Scalar and Vector Potentials

The electromagnetic fields can be expressed in terms of the **scalar potential** $\phi$ and the **vector potential** $\mathbf{A}$ as
$$
\mathbf{B} = \nabla \times \mathbf{A}, \qquad
\mathbf{E} = -\nabla \phi - \frac{\partial \mathbf{A}}{\partial t}.
$$

By substituting these definitions into Maxwell’s equations, the two homogeneous equations,
$$
\nabla \cdot \mathbf{B} = 0, \qquad
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t},
$$
are automatically satisfied.

The remaining two equations,
$$
\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}, \qquad
\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t},
$$
can then be rewritten as wave equations for the potentials.

In the **Lorenz gauge**, defined by
$$
\nabla \cdot \mathbf{A} + \frac{1}{c^2}\frac{\partial \phi}{\partial t} = 0,
$$
the equations for $\phi$ and $\mathbf{A}$ take the simple form
$$
\Box \phi = -\frac{\rho}{\varepsilon_0}, \qquad
\Box \mathbf{A} = -\mu_0 \mathbf{J},
$$
where $\Box = \nabla^2 - \frac{1}{c^2}\frac{\partial^2}{\partial t^2}$ is the d’Alembert operator.

For a detailed discussion of the differential and integral forms of Maxwell's equations, as well as their physical interpretation, see  
**J. D. Jackson, *Classical Electrodynamics*, 3rd edition, Chapter 6.**

## 1.2 Covariant Formulation of Electrodynamics

The covariant formulation of Maxwell's equations expresses the electromagnetic field in a manifestly Lorentz-invariant form.  
We introduce the **four-potential**
$$
A^{\mu} = (\phi, \mathbf{A}),
$$
where $\phi$ is the scalar potential and $\mathbf{A}$ the vector potential.  
The **electromagnetic field tensor** (or Faraday tensor) is defined as
$$
F_{\mu\nu} = \partial_{\mu} A_{\nu} - \partial_{\nu} A_{\mu}.
$$

In this notation, Maxwell's equations reduce to two compact tensor equations:
$$
\partial_{\mu} F^{\mu\nu} = \mu_0 J^{\nu}, \qquad
\partial_{[\alpha} F_{\beta\gamma]} = 0,
$$
where $J^{\nu} = (\rho c, \mathbf{J})$ is the four-current, and square brackets denote antisymmetrization:
$$
\partial_{[\alpha} F_{\beta\gamma]} = 
\frac{1}{3!}\left(
\partial_{\alpha} F_{\beta\gamma} + 
\partial_{\beta} F_{\gamma\alpha} + 
\partial_{\gamma} F_{\alpha\beta}
\right) = 0.
$$

The first equation corresponds to the **inhomogeneous** Maxwell equations, while the second represents the **homogeneous** ones, which follow from the antisymmetry of $F_{\mu\nu}$.

The field tensor can be written explicitly in matrix form as
$$
F_{\mu\nu} =
\begin{pmatrix}
0 & -E_x/c & -E_y/c & -E_z/c \\
E_x/c & 0 & -B_z & B_y \\
E_y/c & B_z & 0 & -B_x \\
E_z/c & -B_y & B_x & 0
\end{pmatrix}.
$$

This formalism makes the Lorentz covariance of electromagnetism explicit: the six independent components of $F_{\mu\nu}$ correspond to the three components of $\mathbf{E}$ and $\mathbf{B}$.

## 1.3 Levi–Civita Tensor, Hodge Dual, and Lorentz Invariants

### 1.3.1 Levi–Civita Tensor

The **Levi–Civita tensor** $\epsilon_{\mu\nu\rho\sigma}$ is the totally antisymmetric rank-4 tensor that generalizes the three-dimensional symbol $\epsilon_{ijk}$ to four-dimensional spacetime.  
It is defined in terms of the **Levi–Civita symbol** $[\,\mu\nu\rho\sigma\,]$, which takes the values

$$
[\,\mu\nu\rho\sigma\,] =
\begin{cases}
+1 & \text{if } (\mu\nu\rho\sigma) \text{ is an even permutation of } (0,1,2,3),\\[3pt]
-1 & \text{if it is an odd permutation},\\[3pt]
0 & \text{if any two indices coincide.}
\end{cases}
$$

In a general curved spacetime with metric determinant $g = \det(g_{\mu\nu})$, the tensor version is
$$
\epsilon_{\mu\nu\rho\sigma} = \sqrt{-g}\,[\,\mu\nu\rho\sigma\,], \qquad
\epsilon^{\mu\nu\rho\sigma} = -\frac{1}{\sqrt{-g}}\,[\,\mu\nu\rho\sigma\,],
$$
where the sign conventions correspond to the metric signature $(-,+,+,+)$.  
In flat spacetime (Minkowski coordinates, $g=-1$),
$$
\epsilon_{0123} = +1, \qquad \epsilon^{0123} = -1.
$$

The Levi–Civita tensor satisfies the contraction identity
$$
\epsilon^{\mu\nu\rho\sigma}\,\epsilon_{\alpha\beta\rho\sigma}
= -2!\,2!\,\delta^{[\mu}_{\alpha}\,\delta^{\nu]}_{\beta}
= -2\!\left(\delta^{\mu}_{\alpha}\delta^{\nu}_{\beta} - \delta^{\mu}_{\beta}\delta^{\nu}_{\alpha}\right),
$$
which is frequently used when manipulating dual tensors.

### 1.3.2 Hodge Dual of the Field Tensor

Given the electromagnetic field tensor $F_{\mu\nu}$, its (left) **Hodge dual** is defined by
$$
{}^{\ast}F_{\mu\nu} \equiv \frac{1}{2}\,\epsilon_{\mu\nu}{}^{\rho\sigma} F_{\rho\sigma}.
$$
With our conventions,
$$
{}^{\ast}({}^{\ast}F)_{\mu\nu} = -F_{\mu\nu}.
$$

### 1.3.3 Scalar (Pseudo)Invariants
There are two independent Lorentz (pseudo)invariants built from $F_{\mu\nu}$:
$$
I_1 \;\equiv\; F_{\mu\nu}F^{\mu\nu},
\qquad
I_2 \;\equiv\; F_{\mu\nu}\,{}^{\ast}F^{\mu\nu}.
$$
In SI units (and with the conventions used earlier where the $F_{\mu\nu}$ matrix contains $\mathbf{E}/c$),
$$
I_1 \;=\; 2\!\left(\,|\mathbf{B}|^2 - \frac{|\mathbf{E}|^2}{c^2}\right),
\qquad
I_2 \;=\; -\,\frac{4}{c}\,\mathbf{E}\!\cdot\!\mathbf{B}.
$$
Useful identities include
$$
{}^{\ast}F_{\mu\nu}\,{}^{\ast}F^{\mu\nu} \;=\; -\,I_1,
\qquad
\frac{1}{4}\,\epsilon^{\mu\nu\rho\sigma}F_{\mu\nu}F_{\rho\sigma}
\;=\; \frac{1}{2}\,I_2.
$$

### 1.3.4 Physical Interpretation
- **Field type (electric-like vs magnetic-like vs null):**
$$ 
  \begin{cases}
  I_1 < 0 &\Rightarrow \text{electric-dominated field (there exists a frame with }\mathbf{B}'=0),\\[3pt]
  I_1 > 0 &\Rightarrow \text{magnetic-dominated field (there exists a frame with }\mathbf{E}'=0),\\[3pt]
  I_1 = 0 = I_2 &\Rightarrow \text{null (radiative) field, e.g. plane wave with }|\mathbf{E}|=c|\mathbf{B}|,\ \mathbf{E}\perp\mathbf{B}.
  \end{cases}
$$
- **Mutual orientation of $\mathbf{E}$ and $\mathbf{B}$:** $I_2 \propto \mathbf{E}\!\cdot\!\mathbf{B}$. Thus $I_2=0$ iff $\mathbf{E}\perp\mathbf{B}$ in every inertial frame (no frame exists where both vanish unless the field is trivial).
- **Invariance under Lorentz transformations:** $I_1$ and $I_2$ have the same value in all inertial frames; they classify the electromagnetic field independent of observer.
- **Parity properties:** $I_1$ is a true scalar; $I_2$ is a pseudoscalar (changes sign under improper transformations that reverse orientation).


For detailed derivations and conventions, see  
**J. D. Jackson, *Classical Electrodynamics*, 3rd ed., Chapter 11;**  
**L. D. Landau and E. M. Lifshitz, *The Classical Theory of Fields*, §§23–25 (Ch. 3);**  
**C. W. Misner, K. S. Thorne, J. A. Wheeler, *Gravitation*, Chapter 4.**

## 1.4 The Energy–Momentum Tensor of the Electromagnetic Field

In special relativity, the energy and momentum carried by the electromagnetic field are encoded in the **stress–energy tensor**
$$
T^{\mu\nu} = \frac{1}{\mu_0}\left( 
F^{\mu\alpha} F^{\nu}{}_{\alpha} 
- \frac{1}{4}\, g^{\mu\nu} F_{\alpha\beta} F^{\alpha\beta}
\right).
$$

This tensor is symmetric, traceless in vacuum, and satisfies the local conservation law
$$
\partial_{\mu} T^{\mu\nu} = - F^{\nu\lambda} J_{\lambda}.
$$
In regions where $J_{\lambda}=0$, this reduces to $\partial_{\mu} T^{\mu\nu} = 0$, expressing the conservation of energy and momentum of the free electromagnetic field.

In the laboratory (3+1) decomposition, the components of $T^{\mu\nu}$ can be written as
$$
T^{00} = \frac{1}{2}\left(\varepsilon_0 |\mathbf{E}|^2 + \frac{1}{\mu_0}|\mathbf{B}|^2\right), \qquad
T^{0i} = \frac{1}{\mu_0 c} (\mathbf{E} \times \mathbf{B})^i,
$$

where $T^{00}$ represents the **energy density** and $T^{0i}$ the **Poynting vector**, corresponding to the flux of electromagnetic energy.

The tensor $T^{ij}$ (spatial components) represents the **Maxwell stress tensor**, describing momentum flux and field pressure:
$$
T^{ij} = \varepsilon_0 \left( E^i E^j + c^2 B^i B^j - \frac{1}{2}\delta^{ij} (|\mathbf{E}|^2 + c^2 |\mathbf{B}|^2) \right).
$$

In general relativity, $T^{\mu\nu}$ serves as the source term in Einstein’s field equations,
$$
G^{\mu\nu} = \frac{8\pi G}{c^4} T^{\mu\nu},
$$
linking spacetime curvature to the distribution of electromagnetic energy and momentum.

For a detailed derivation and discussion of the physical interpretation of $T^{\mu\nu}$, see  
**C. W. Misner, K. S. Thorne, and J. A. Wheeler, *Gravitation*, Chapter 5.**

## 2. Gravitation

## 2.1 The Metric, Connection, and Field Equations

General Relativity describes gravitation as the manifestation of spacetime curvature produced by energy and momentum.  
The fundamental geometric quantity is the **metric tensor** $g_{\mu\nu}$, which defines distances and causal structure through the line element
$$
ds^2 = g_{\mu\nu}\,dx^{\mu} dx^{\nu}.
$$

### 2.1.1 Christoffel Symbols and Covariant Derivative
Parallel transport and differentiation are expressed using the **Levi–Civita connection**, defined in terms of the metric as
$$
\Gamma^{\rho}_{\ \mu\nu} = \frac{1}{2} g^{\rho\sigma}
\left( \partial_{\mu} g_{\nu\sigma} + \partial_{\nu} g_{\mu\sigma}
      - \partial_{\sigma} g_{\mu\nu} \right).
$$
The **covariant derivative** of a vector field $V^{\mu}$ is then
$$
\nabla_{\nu} V^{\mu} = \partial_{\nu} V^{\mu} + \Gamma^{\mu}_{\ \nu\rho} V^{\rho}.
$$

### 2.1.2 Riemann Curvature Tensor
The curvature of spacetime is encoded in the **Riemann tensor**, which measures the non-commutativity of covariant derivatives:
$$
(\nabla_{\mu}\nabla_{\nu} - \nabla_{\nu}\nabla_{\mu}) V^{\rho}
= R^{\rho}_{\ \sigma\mu\nu} V^{\sigma}.
$$
In component form,
$$
R^{\rho}_{\ \sigma\mu\nu} =
\partial_{\mu}\Gamma^{\rho}_{\ \nu\sigma}
- \partial_{\nu}\Gamma^{\rho}_{\ \mu\sigma}
+ \Gamma^{\rho}_{\ \mu\lambda}\Gamma^{\lambda}_{\ \nu\sigma}
- \Gamma^{\rho}_{\ \nu\lambda}\Gamma^{\lambda}_{\ \mu\sigma}.
$$

By contraction, one obtains the **Ricci tensor** and **Ricci scalar**:
$$
R_{\mu\nu} = R^{\rho}_{\ \mu\rho\nu}, \qquad
R = g^{\mu\nu} R_{\mu\nu}.
$$

### 2.1.3 Einstein Field Equations
The geometry of spacetime is related to the energy–momentum tensor $T_{\mu\nu}$ through the **Einstein field equations**
$$
G_{\mu\nu} = \frac{8\pi G}{c^{4}}\,T_{\mu\nu},
$$
where the Einstein tensor is defined as
$$
G_{\mu\nu} = R_{\mu\nu} - \frac{1}{2} R\, g_{\mu\nu}.
$$

In vacuum, where $T_{\mu\nu}=0$, these equations reduce to
$$
R_{\mu\nu} = 0,
$$
and the dynamics of curvature are entirely described by the Riemann (or equivalently the Weyl) tensor.

For detailed derivations and discussion, see  
**C. W. Misner, K. S. Thorne, and J. A. Wheeler, *Gravitation*, Chapters 13-17**,  
**L. D. Landau and E. M. Lifshitz, *The Classical Theory of Fields*, Chapter 11.**

## 2.1.4 Causal Structure and Spacetime Classification

At each point in spacetime, the **metric tensor** $g_{\mu\nu}$ defines the inner product between two vectors $A^{\mu}$ and $B^{\mu}$ as
$$
A \cdot B = g_{\mu\nu} A^{\mu} B^{\nu}.
$$
According to the sign of this scalar product, one distinguishes three types of vectors:

$$
\begin{cases}
A_{\mu}A^{\mu} < 0 &\Rightarrow \text{timelike},\\[4pt]
A_{\mu}A^{\mu} = 0 &\Rightarrow \text{null (lightlike)},\\[4pt]
A_{\mu}A^{\mu} > 0 &\Rightarrow \text{spacelike}.
\end{cases}
$$

Timelike vectors correspond to worldlines of massive particles, null vectors to light rays, and spacelike vectors to directions that cannot be reached by any causal signal.  
This classification extends to hypersurfaces and tensor fields: for example, a **null hypersurface** has a normal vector that is itself null.

## 2.1.5 Levi–Civita Tensor in Curved Spacetime

The **Levi–Civita tensor** $\epsilon_{\mu\nu\rho\sigma}$ generalizes the completely antisymmetric symbol to curved spacetime, incorporating the metric determinant:
$$
\epsilon_{\mu\nu\rho\sigma} = \sqrt{-g}\,[\mu\nu\rho\sigma], \qquad
\epsilon^{\mu\nu\rho\sigma} = -\frac{1}{\sqrt{-g}}\,[\mu\nu\rho\sigma],
$$
where $[\mu\nu\rho\sigma]$ is the permutation symbol and $g = \det(g_{\mu\nu})$.  
It satisfies the identity
$$
\epsilon^{\mu\nu\rho\sigma}\,\epsilon_{\alpha\beta\rho\sigma}
= -2\left(\delta^{\mu}_{\alpha}\delta^{\nu}_{\beta} - \delta^{\mu}_{\beta}\delta^{\nu}_{\alpha}\right).
$$

The Levi–Civita tensor allows one to define dual tensors in curved spacetime, such as the dual of the Riemann or Weyl tensor.

## 2.1.6 The Weyl Tensor and Its Dual

The **Riemann tensor** can be decomposed into trace and trace-free parts.  
The trace-free part is the **Weyl tensor**, which encodes the purely gravitational degrees of freedom (tidal forces and gravitational radiation):
$$
C_{\alpha\beta\gamma\delta} =
R_{\alpha\beta\gamma\delta}
- \frac{1}{2}\left(
g_{\alpha[\gamma}R_{\delta]\beta} - g_{\beta[\gamma}R_{\delta]\alpha}
\right)
+ \frac{1}{6} R\, g_{\alpha[\gamma}g_{\delta]\beta}.
$$
By construction,
$$
C^{\mu}{}_{\nu\mu\rho} = 0, \qquad g^{\mu\rho} C_{\mu\nu\rho\sigma} = 0.
$$

The **dual Weyl tensor** is defined using the Levi–Civita tensor:
$$
{}^{\ast}C_{\alpha\beta\gamma\delta}
= \frac{1}{2}\,\epsilon_{\alpha\beta}{}^{\mu\nu}\,C_{\mu\nu\gamma\delta},
$$
and satisfies
$$
{}^{\ast}({}^{\ast}C)_{\alpha\beta\gamma\delta} = -\,C_{\alpha\beta\gamma\delta}.
$$

The Weyl tensor determines the part of curvature that is independent of local matter sources ($T_{\mu\nu}$).  
In vacuum, where $R_{\mu\nu}=0$, the Riemann and Weyl tensors coincide:
$$
R_{\alpha\beta\gamma\delta} = C_{\alpha\beta\gamma\delta}.
$$

## 2.2 Timelike Decomposition of the Weyl Tensor

When a timelike unit vector field $u^{\alpha}$ is chosen to represent a family of observers 
($u_{\alpha}u^{\alpha}=-1$), the Weyl tensor can be decomposed into its **electric** and **magnetic** parts, 
in complete analogy with the electromagnetic field tensor.

### 2.2.1 Electric and Magnetic Parts of the Weyl Tensor
The **electric** and **magnetic** parts of the Weyl tensor are defined as
$$
E_{\alpha\beta} = C_{\alpha\mu\beta\nu} u^{\mu} u^{\nu}, \qquad
B_{\alpha\beta} = {}^{\ast}C_{\alpha\mu\beta\nu} u^{\mu} u^{\nu}
= \frac{1}{2}\,\epsilon_{\alpha\mu}{}^{\rho\sigma} C_{\rho\sigma\beta\nu} u^{\mu} u^{\nu}.
$$
Both tensors are symmetric, trace-free, and orthogonal to $u^{\alpha}$:
$$
E_{\alpha\beta} = E_{(\alpha\beta)}, \quad
B_{\alpha\beta} = B_{(\alpha\beta)}, \quad
E^{\alpha}{}_{\alpha} = B^{\alpha}{}_{\alpha} = 0, \quad
E_{\alpha\beta}u^{\beta} = B_{\alpha\beta}u^{\beta} = 0.
$$

Physically, $E_{\alpha\beta}$ describes **tidal gravitational accelerations**, while $B_{\alpha\beta}$ encodes 
**frame-dragging** and the **gravitomagnetic** part of curvature.

### 2.2.2 Lorentz Invariants of the Weyl Field

The scalar invariants of the Weyl tensor are defined by
$$
J_1 = \frac{1}{8}\, C_{\alpha\beta\gamma\delta} C^{\alpha\beta\gamma\delta}, \qquad
J_2 = \frac{1}{16}\, C_{\alpha\beta\gamma\delta}\, {}^{\ast}C^{\alpha\beta\gamma\delta}.
$$
In terms of the electric and magnetic parts,
$$
J_1 = E_{\alpha\beta}E^{\alpha\beta} - B_{\alpha\beta}B^{\alpha\beta}, \qquad
J_2 = 2\, E_{\alpha\beta} B^{\alpha\beta}.
$$

These invariants are the gravitational analogues of the electromagnetic scalars
$$
I_1 = F_{\mu\nu}F^{\mu\nu}, \qquad
I_2 = F_{\mu\nu}\,{}^{\ast}F^{\mu\nu},
$$
and they play an analogous role:
- $J_1$ measures the relative dominance of the "gravitoelectric" versus "gravitomagnetic" parts.
- $J_2$ encodes their mutual orientation and parity properties.
- A **null (radiative) spacetime** satisfies $J_1 = 0 = J_2$, corresponding to Petrov type N.

### 2.2.3 The Bel–Robinson Tensor

The **Bel–Robinson tensor** $T_{\alpha\beta\gamma\delta}$ is the gravitational analogue of the electromagnetic stress–energy tensor.  
In vacuum, it is defined by
$$
T_{\alpha\beta\gamma\delta}
= C_{\alpha\mu\gamma\nu} C_{\beta}{}^{\mu}{}_{\delta}{}^{\nu}
+ {}^{\ast}C_{\alpha\mu\gamma\nu}\,{}^{\ast}C_{\beta}{}^{\mu}{}_{\delta}{}^{\nu}.
$$
It is completely symmetric, trace-free, and divergence-free:
$$
\nabla^{\alpha} T_{\alpha\beta\gamma\delta} = 0.
$$

For a timelike observer $u^{\alpha}$, one defines the **super-energy density** and **super-Poynting vector** as
$$
U = T_{\alpha\beta\gamma\delta} u^{\alpha}u^{\beta}u^{\gamma}u^{\delta}
= E_{\alpha\beta}E^{\alpha\beta} + B_{\alpha\beta}B^{\alpha\beta},
$$
$$
P_{\mu} = T_{\alpha\beta\gamma\delta} u^{\alpha}u^{\beta}u^{\gamma}h^{\delta}{}_{\mu}
= 2\,\epsilon_{\mu\nu\rho}\,E^{\nu}{}_{\sigma} B^{\rho\sigma},
$$
where $h_{\mu\nu} = g_{\mu\nu} + u_{\mu}u_{\nu}$ projects orthogonally to $u^{\alpha}$.
The quantity $U$ measures the **super-energy density** of the gravitational field, while $P_{\mu}$ describes the **flux of super-energy** — the gravitational analogue of the electromagnetic Poynting vector.

### 2.2.4 Summary of the Timelike Formalism

In this approach, the analysis proceeds as follows:

1. **Start from the metric** $g_{\mu\nu}$,  
2. **Compute the Weyl tensor** $C_{\alpha\beta\gamma\delta}$,  
3. **Decompose** it into $E_{\alpha\beta}$ and $B_{\alpha\beta}$ using $u^{\alpha}$,  
4. **Construct the invariants** $J_1, J_2$ to determine the algebraic field type,  
5. **Form the Bel–Robinson tensor** $T_{\alpha\beta\gamma\delta}$,  
6. **Compute** the associated super-energy density $U$ and flux $P^{\alpha}$.

This formalism establishes a complete analogy between electromagnetic and gravitational fields in the presence of a timelike observer.

For references and derivations, see  
**G. F. R. Ellis and R. Maartens, *Relativistic Cosmology*, Chapter 4**,  
**R. Maartens, Gravito-electromagnetism**.

## 2.3 Null (Newman–Penrose) Formalism for Radiation

In radiative regimes (far zone of asymptotically flat spacetimes), the Weyl invariants vanish,
$$
J_1 = \tfrac18\, C_{\alpha\beta\gamma\delta} C^{\alpha\beta\gamma\delta} = 0,
\qquad
J_2 = \tfrac1{16}\, C_{\alpha\beta\gamma\delta} {}^{\ast}C^{\alpha\beta\gamma\delta} = 0,
$$
characterizing a **null** curvature field (Petrov type N at leading order).

### 2.3.1 Null tetrad and NP scalars 
Choose a complex null tetrad $(l^a, n^a, m^a, \bar m^a)$ with signature $(-,+,+,+)$ and inner products

$$
\ell\!\cdot\!n = -1, \qquad m\!\cdot\!\bar m = 1,
\qquad \text{all other products zero}.
$$
The five Newman–Penrose (NP) Weyl scalars are
$$
\begin{aligned}
\psi_0 &= - C_{abcd}\,\ell^a m^b \ell^c m^d,\\
\psi_1 &= - C_{abcd}\,\ell^a n^b \ell^c m^d,\\
\psi_2 &= - C_{abcd}\,\ell^a m^b \bar m^c n^d,\\
\psi_3 &= - C_{abcd}\,\ell^a n^b \bar m^c n^d,\\
\psi_4 &= - C_{abcd}\,n^a \bar m^b n^c \bar m^d.
\end{aligned}
$$
The $10$ real independent components of $C_{abcd}$ are equivalently encoded by the $5$ complex scalars $(\psi_0,\dots,\psi_4)$ once a null tetrad is fixed.

*Source:* S. Chandrasekhar, *The Mathematical Theory of Black Holes*, Ch. 1.

### 2.3.2 Peeling theorem and physical meaning 
Along outgoing null geodesics to future null infinity $\mathscr{I}^+$ (Bondi radius $r\to\infty$), the Weyl scalars “peel” as
$$
\psi_0 = \mathcal{O}(r^{-5}),\quad
\psi_1 = \mathcal{O}(r^{-4}),\quad
\psi_2 = \mathcal{O}(r^{-3}),\quad
\psi_3 = \mathcal{O}(r^{-2}),\quad
\psi_4 = \mathcal{O}(r^{-1}).
$$

$$
\begin{array}{c|c|l}
\text{Scalar} & \text{Falloff (outgoing)} & \text{Interpretation (heuristic)}\\ \hline
\psi_4 & r^{-1} & \text{Outgoing transverse radiation (wave strain at }\mathscr{I}^+\text{)}\\
\psi_3 & r^{-2} & \text{Near-zone longitudinal/transverse–longitudinal fields}\\
\psi_2 & r^{-3} & \text{Coulombic/monopole mass aspect (stationary part)}\\
\psi_1 & r^{-4} & \text{Longitudinal “dipole-like” tail}\\
\psi_0 & r^{-5} & \text{Ingoing radiation near }\mathscr{I}^-\text{ (in suitable tetrads)}
\end{array}
$$

In **stationary** examples one expects only the Coulombic piece ($\psi_2\neq 0$, $\psi_{0,1,3,4}=0$ in an adapted tetrad).  
In **radiative far zones**, one can choose an outgoing tetrad with $\psi_4$ dominant and $\psi_{0,1,2,3}$ subleading by the peeling hierarchy.  
This aligns with the **Petrov classification**: type D for stationary Coulomb fields (e.g. Kerr), type N for pure radiation.

*Sources*: J. Stewart, *Advanced General Relativity, esp. §3.5*; *J. Gómez-Lobo (appendix on Petrov types)*.

### 2.3.3 Gravitational Plane Waves and Quadrupole Radiation

In the weak–field limit, gravitational waves are treated as small perturbations on flat spacetime,

$$
g_{\mu\nu} = \eta_{\mu\nu} + h_{\mu\nu}, \qquad |h_{\mu\nu}| \ll 1,
$$

and satisfy the linearized Einstein equations in the **transverse–traceless (TT) gauge**:

$$
\partial^\mu h_{\mu\nu}^{\mathrm{TT}} = 0, \qquad h^{\mathrm{TT}} = 0, \qquad h_{0\mu}^{\mathrm{TT}} = 0.
$$

Here $ h_{ij}^{\mathrm{TT}} $ (Latin indices for spatial components) denotes the **transverse–traceless metric perturbation**, which represents the physical, gauge‐invariant degrees of freedom of the gravitational field.

#### 2.3.3.1 Plane wave form

For a plane wave propagating in the $+z$ direction, the TT metric takes the standard form

$$
ds^2 = -c^2 dt^2 + [1 + h_+(t - z/c)]\,dx^2 + [1 - h_+(t - z/c)]\,dy^2
+ 2h_\times(t - z/c)\,dx\,dy + dz^2.
$$

The two functions $h_+(u)$ and $h_\times(u)$, with $u = t - z/c$, represent the two independent polarization modes of the wave.  
They correspond to the **quadrupole radiation** emitted by a time‐varying, nonspherical mass distribution.

#### 2.3.3.2 Quadrupole formula

In the slow‐motion, weak‐field limit (far from the source), the TT part of the metric perturbation is given by the **quadrupole formula**

$$
h_{ij}^{\mathrm{TT}}(t, \mathbf{x})
= \frac{2G}{c^4 r}\,
\ddot{Q}_{ij}^{\mathrm{TT}}(t - r/c),
$$

where $ Q_{ij} $ is the **mass quadrupole moment tensor**

$$
Q_{ij}(t) = \int \rho(t,\mathbf{x})
\left(x_i x_j - \tfrac{1}{3}\,\delta_{ij}\,r^2\right) d^3x,
$$

and the TT projection removes unphysical longitudinal components.  
This expression shows that gravitational waves are produced by **time‐varying quadrupole (and higher multipole) moments**, with amplitudes falling off as $ 1/r $.

#### 2.3.3.3 Relation to ψ₄

Choosing an **outgoing null tetrad** $(\ell^a,n^a,m^a,\bar m^a)$ aligned with the propagation direction, the only nonvanishing Weyl scalar in the far‐field limit is

$$
\psi_4 \sim \left( \ddot h_+ - i\,\ddot h_\times \right).
$$

Thus, ψ₄ encodes the curvature component directly proportional to the second time derivative of the TT strain.

#### 2.3.3.4 Energy flux and ψ₄ equivalence

In the Isaacson short–wavelength limit, the averaged energy flux carried by the gravitational wave is

$$
\langle S \rangle = \frac{c^3}{32\pi G} 
\left\langle \dot h_+^{\,2} + \dot h_\times^{\,2} \right\rangle.
$$

Using the relation between $h_{+,\times}$ and ψ₄,

$$
\dot h_+ - i\,\dot h_\times \;\; \propto \;\; - \int^t \psi_4 \, dt',
$$

we find, up to overall constants and conventions,

$$
\langle S \rangle \;\propto\; \big\langle\,|\int^{t}\psi_4 dt'|^2\,\big\rangle.
$$

Hence, in the **radiative far zone**, ψ₄ not only describes the curvature of outgoing waves but also carries the same **energy flux information** as the Isaacson stress–energy tensor.

R. Isaacson, *Phys. Rev.* (1968) I & II.

### 2.3.4 Spinor form, Bel–Robinson tensor, and NP components 
In 2–spinor notation, the Weyl tensor is represented by the totally symmetric **Weyl spinor** $\Psi_{ABCD}$.  
Relative to a normalized spinor dyad $(o^A, \iota^A)$, one constructs the corresponding **null tetrad** of vectors by the standard correspondence between spinor indices and vector indices:

$$
\ell^a = o^A \bar{o}^{A'}, \qquad
n^a = \iota^A \bar{\iota}^{A'}, \qquad
m^a = o^A \bar{\iota}^{A'}, \qquad
\bar m^a = \iota^A \bar{o}^{A'}.
$$

These vectors satisfy the normalization conditions

$$
\ell \!\cdot\! n = -1, \qquad m \!\cdot\! \bar m = 1, \qquad
\text{all other scalar products vanish.}
$$

With these conventions, the **Weyl spinor** $\Psi_{ABCD}$ encodes the self-dual part of the Weyl tensor, and its contractions with $(o^A,\iota^A)$ give the **five Newman–Penrose scalars**:

$$
\psi_0 = \Psi_{ABCD}\,o^A o^B o^C o^D, \qquad
\psi_1 = \Psi_{ABCD}\,o^A o^B o^C \iota^D, \qquad
\psi_2 = \Psi_{ABCD}\,o^A o^B \iota^C \iota^D,
$$

$$
\psi_3 = \Psi_{ABCD}\,o^A \iota^B \iota^C \iota^D, \qquad
\psi_4 = \Psi_{ABCD}\,\iota^A \iota^B \iota^C \iota^D.
$$

Finally, the **Bel–Robinson tensor** takes the compact spinor form

$$
T_{AA'BB'CC'DD'} = \Psi_{ABCD}\,\bar\Psi_{A'B'C'D'},
$$

which is the natural gravitational analogue of the electromagnetic stress–energy tensor $T^{\mu\nu}_{\mathrm{EM}}$.

*Sources:*  
R. Penrose and W. Rindler, *Spinors and Space–Time*, Vol. 1;  
W. Rindler, *Relativity: Special, General, and Cosmology*, §4.8;  
J. Stewart, *Advanced General Relativity*, §2.6.

### 2.3.5 Null tetrad components of the Bel–Robinson tensor

In the Newman–Penrose formalism, the components of $T_{abcd}$ can be expressed in terms of the Weyl scalars.  
For an outgoing null tetrad $(\ell^a, n^a, m^a, \bar m^a)$, the leading contractions are

$$
T_{nnnn} = 4\,|\psi_4|^2, \qquad
T_{\ell\ell\ell\ell} = 4\,|\psi_0|^2,
$$

which represent, respectively, the **outgoing** and **ingoing** radiative super–energy fluxes. Other terms consist of combinations of the NP scalars.

Thus, in the **radiative (type N)** regime, only $\psi_4$ survives asymptotically, and the super–energy flux is carried by

$$
T_{nnnn} \propto |\psi_4|^2.
$$

For reference,

$$
\text{Coulombic:} \quad \psi_2 \sim \frac{\text{mass aspect}}{r^3}, \qquad
\text{Radiative:} \quad \psi_4 \sim \ddot h_+ - i\,\ddot h_\times \sim \mathcal{O}(r^{-1}).
$$

*Sources:*  
W. Rindler, *Relativity: Special, General, and Cosmology*, §4.8 (spinors), eqs. (4,12,15) and (4,11,9);  
J. Stewart, *Advanced General Relativity*, eq. (2.6.9).

### Workflow in the null/radiative formalism
1. Choose an asymptotically outgoing null tetrad $(\ell^a,n^a,m^a,\bar m^a)$.  
2. Compute the Weyl scalars $\psi_i$ from $C_{abcd}$.  
3. Use peeling to identify Coulombic vs radiative content (expect $\psi_4$ dominance at $\mathscr{I}^+$).  
4. Relate $\psi_4$ to the measured strain $(h_+,h_\times)$ in the Isaacson limit.  
5. Build the Bel–Robinson tensor $T_{abcd}$; radiative flux $\propto \psi_4\bar\psi_4$.

Maggiore Vol. 1; Thorne GW lectures; Rindler §4.8.

# Part II – Computational Examples

In this part we apply the theoretical framework developed above to a series of representative spacetimes of increasing physical and algebraic complexity.  
Each example will be analyzed in a dedicated notebook, where we will compute the Weyl tensor, the Newman–Penrose scalars, and the related curvature invariants.  
The goal is to illustrate, in concrete settings, how the mathematical structure of the field encodes its physical content.

### 3.1 Conformally Flat Spacetime

A spacetime that is locally conformal to Minkowski space, possessing no free gravitational field.  
This serves as the simplest possible test case and a reference for verifying that the formalism correctly yields a vanishing Weyl tensor.

### 3.2 Schwarzschild Metric

A static, spherically symmetric vacuum solution representing the external field of a non-rotating mass.  
This metric illustrates a purely Coulombic gravitational field with no radiative degrees of freedom, providing a natural baseline for comparison with rotating and time-dependent cases.

### 3.3 Kerr Metric

A stationary, axisymmetric solution describing the geometry around a rotating mass.  
It introduces gravito-magnetic effects such as frame dragging and shows how rotation modifies the algebraic structure of the curvature while maintaining stationarity.

### 3.4 Plane Gravitational Wave on Minkowski Spacetime

A simple model of a weak gravitational wave propagating in flat space.  
This example represents a purely radiative spacetime and provides a direct link between the curvature description (via the Weyl tensor) and the observable wave polarizations $ h_+ $ and $ h_\times $.

### 3.5 Quasi-Circular Binary in the Far Zone

An astrophysically relevant configuration producing gravitational radiation through orbital motion.  
In the far zone, the metric can be approximated by a perturbation on flat or asymptotically flat spacetime, allowing us to study the asymptotic behavior of the curvature and the relation between emitted radiation and the underlying source dynamics.

These five cases collectively trace the progression from **non-radiative** to **fully radiative** gravitational fields.  
They will form the computational backbone of this project, each treated in its own notebook following the common workflow:
metric → curvature → Newman–Penrose scalars → invariants → physical interpretation.
